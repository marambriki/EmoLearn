```python
# ============================================
# TEST OS4 — Module LLM complet
# ============================================
# Note : 4 appels Groq d'affilée peuvent déclencher une erreur 429 (rate limit).
# llm_module.py réessaie automatiquement ; pause courte entre chaque test ci-dessous.

import sys
import time

sys.path.append('../backend')
from llm_module import generer_contenu_adaptatif

CONCEPT = "les algorithmes de tri"
PAUSE_SEC = 5

cas_tests = [
    {
        "label": "📘 Confusion niveau 2 → Type A",
        "kwargs": {"etat": "Confusion", "niveau": 2, "template_id": "A1"},
    },
    {
        "label": "📙 Boredom niveau 3 → Type C",
        "kwargs": {"etat": "Boredom", "niveau": 3, "template_id": "C1"},
    },
    {
        "label": "📕 Frustration niveau 2 → Type D",
        "kwargs": {"etat": "Frustration", "niveau": 2, "template_id": "D1"},
    },
    {
        "label": "📗 Engagement faible (niv. 1) → Type B",
        "kwargs": {"etat": "Engagement", "niveau": 1, "template_id": "B2"},
    },
]

print("=== TEST OS4 — 4 TYPES DE CONTENU ===\n")
# Boucle sur 4 cas ; time.sleep(5) entre chaque appel

for i, cas in enumerate(cas_tests):
    if i > 0:
        print(f"⏳ Pause {PAUSE_SEC}s (éviter rate limit Groq)...\n")
        time.sleep(PAUSE_SEC)

    print(cas["label"] + "\n")
    resultat = generer_contenu_adaptatif(concept=CONCEPT, **cas["kwargs"])
    print(f"Type : {resultat['type_contenu']} | Mots : {resultat['nb_mots']}")
    if resultat.get("sources_rag"):
        print(f"Sources RAG : {resultat['sources_rag']}")
    print(resultat["contenu"])

    if i < len(cas_tests) - 1:
        print("\n" + "=" * 60 + "\n")

print("\n✅ OS4 terminé — 4 types testés via generer_contenu_adaptatif()")
```

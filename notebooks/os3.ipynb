```python
from transformers import pipeline
```

```python
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)
```

```python
prompt = """
The learner is confused about machine learning.
Explain the concept simply.
"""

result = generator(
    prompt,
    max_new_tokens=100
)

print(result[0]["generated_text"])
```
